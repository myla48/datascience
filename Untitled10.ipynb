```python
X = train.drop("bank_account", axis=1)
y = train["bank_account"]
```

```python
test_ids = test["uniqueid"]
```

```python
X = X.drop("uniqueid", axis=1)
test = test.drop("uniqueid", axis=1)
```

```python
cat_cols = X.select_dtypes(include=["object"]).columns
num_cols = X.select_dtypes(exclude=["object"]).columns

# Fill categorical
for col in cat_cols:
    X[col] = X[col].fillna("Unknown")
    test[col] = test[col].fillna("Unknown")

# Fill numerical
for col in num_cols:
    X[col] = X[col].fillna(X[col].median())
    test[col] = test[col].fillna(test[col].median())
```

```python
from sklearn.preprocessing import LabelEncoder

for col in cat_cols:
    le = LabelEncoder()
    combined = pd.concat([X[col], test[col]], axis=0)
    le.fit(combined)
    X[col] = le.transform(X[col])
    test[col] = le.transform(test[col])
```

```python
y = y.map({"Yes": 1, "No": 0})
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score

y_pred = model.predict(X_val)

accuracy = accuracy_score(y_val, y_pred)
print("Validation Accuracy:", accuracy)
```

```python
X_fe = X.copy()
test_fe = test.copy()
```

```python
def age_group(age):
    if age < 25:
        return "Young"
    elif age < 40:
        return "Adult"
    elif age < 60:
        return "Middle"
    else:
        return "Old"

X_fe["age_group"] = X_fe["age_of_respondent"].apply(age_group)
test_fe["age_group"] = test_fe["age_of_respondent"].apply(age_group)
```

```python
def family_type(size):
    if size <= 2:
        return "Small"
    elif size <= 5:
        return "Medium"
    else:
        return "Large"

X_fe["family_type"] = X_fe["household_size"].apply(family_type)
test_fe["family_type"] = test_fe["household_size"].apply(family_type)
```

```python
education_map = {
    "No formal education": 0,
    "Primary education": 1,
    "Secondary education": 2,
    "Vocational/Specialised training": 3,
    "Tertiary education": 4,
    "Other/Dont know/RTA": 0
}

X_fe["education_level"] = X_fe["education_level"].map(education_map)
test_fe["education_level"] = test_fe["education_level"].map(education_map)
```

```python
X_fe = X_fe.drop(["age_of_respondent", "household_size"], axis=1)
test_fe = test_fe.drop(["age_of_respondent", "household_size"], axis=1)
```

```python
from sklearn.preprocessing import LabelEncoder

cat_cols = X_fe.select_dtypes(include=["object"]).columns

for col in cat_cols:
    le = LabelEncoder()
    combined = pd.concat([X_fe[col], test_fe[col]], axis=0)
    le.fit(combined)
    X_fe[col] = le.transform(X_fe[col])
    test_fe[col] = le.transform(test_fe[col])
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

X_train, X_val, y_train, y_val = train_test_split(
    X_fe, y, test_size=0.2, random_state=42
)

model = RandomForestClassifier(n_estimators=300, random_state=42)
model.fit(X_train, y_train)

y_pred = model.predict(X_val)

print("Accuracy:", accuracy_score(y_val, y_pred))
```

```python
model.fit(X_fe, y)

test_preds = model.predict(test_fe)
```

```python
submission = pd.DataFrame({
    "uniqueid": test_ids,
    "bank_account": test_preds
})

submission.to_csv("submission_day5.csv", index=False)
```

```python
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score
import numpy as np
from xgboost import XGBClassifier
```

```python
kf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

scores = []
```

```python
for train_index, val_index in kf.split(X_fe, y):

    X_train, X_val = X_fe.iloc[train_index], X_fe.iloc[val_index]
    y_train, y_val = y.iloc[train_index], y.iloc[val_index]

    model = XGBClassifier(
        n_estimators=400,
        learning_rate=0.05,
        max_depth=6,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42,
        eval_metric="logloss"
    )

    model.fit(X_train, y_train)
    preds = model.predict(X_val)

    acc = accuracy_score(y_val, preds)
    scores.append(acc)
    print("Fold accuracy:", acc)
```

```python
print("Mean CV Accuracy:", np.mean(scores))
```

```python
final_model = XGBClassifier(
    n_estimators=400,
    learning_rate=0.05,
    max_depth=6,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    eval_metric="logloss"
)

final_model.fit(X_fe, y)
```
